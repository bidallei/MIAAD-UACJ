### **Ejercicio 1:** Utiliza el texto crudo (sin procesar)
* Crear un CountVectorizer y un TfidfVectorizer.
* Aplícalos al corpus y transformar los textos.
* Mostrar:
  * Dimensiones de la matriz resultante.
  * Primeros 10 tokens y sus índices.
* Comparar las diferencias entre ambas matrices (CountVectorizer y TfidfVectorizer), mencionando de manera general las diferencias observadas.

```python
# Importamos las librerías
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

# Creamos los vectorizadores
count_vectorizer = CountVectorizer()
tfidf_vectorizer = TfidfVectorizer()

# Aplicamos al corpus y transformamos los textos
matriz_count = count_vectorizer.fit_transform(corpus_es)
matriz_tfidf = tfidf_vectorizer.fit_transform(corpus_es)

# Mostramos las dimensiones de las matrices
print("=== Dimensiones de las matrices ===")
print(f"Forma de la matriz CountVectorizer: {matriz_count.shape}")
print(f"Forma de la matriz TfidfVectorizer: {matriz_tfidf.shape}\n")

# Mostramos los primeros 10
print("=== Primeros 10 tokens y sus índices ===")
vocabulario = count_vectorizer.vocabulary_
primeros_10 = sorted(vocabulario.items(), key=lambda x: x[1])[:10]

for palabra, indice in primeros_10:
    print(f"Índice {indice}: {palabra}")

print("\n")

df_count = pd.DataFrame(
    matriz_count.toarray(),
    columns=count_vectorizer.get_feature_names_out()
)

df_tfidf = pd.DataFrame(
    matriz_tfidf.toarray(),
    columns=tfidf_vectorizer.get_feature_names_out()
)

print("=== Representación CountVectorizer (Muestra de 5 filas x 10 columnas) ===")
print(df_count.iloc[:5, :10], "\n")

print("=== Representación TfidfVectorizer (Muestra de 5 filas x 10 columnas) ===")
print(df_tfidf.iloc[:5, :10], "\n")
```

**Observaciones**
* La matriz de CountVectorizer contiene números enteros que significan el número de veces que aparece cada palabra.
* La matriz de TfidfVectorizer contiene números decimales que representan el peso de cada palabra.


### **Ejercicio 2:** Preprocesa el texto eliminando los acentos y stopwords, así como convirtiéndolo a minúsculas.
* Realiza las mismas actividades que en el ejercicio anterior.
* Compara las matrices del ejercicio anterior con las matrices que obtuviste en este ejercicio, mencionando de manera general las diferencias observadas.








```python
import re
import unicodedata
import nltk
from nltk.corpus import stopwords

nltk.download('stopwords')
stop_words_es = set(stopwords.words('spanish'))

# Hacemos una función para normalizar el texto
def normalizar_texto(texto):
    # Convertimos a minúsculas
    texto = texto.lower()

    # Eliminamos stopwords
    palabras = texto.split()
    palabras_filtradas = [p for p in palabras if p not in stop_words_es]
    texto = ' '.join(palabras_filtradas)

    # Eliminamos acentos
    texto = ''.join(c for c in unicodedata.normalize('NFD', texto)
                    if unicodedata.category(c) != 'Mn')

    # Eliminamos URL's, correos, menciones
    texto = re.sub(r'https?://\S+|www\.\S+', '', texto)
    texto = re.sub(r'\b[\w\.-]+@[\w\.-]+\.\w{2,4}\b', '', texto)
    texto = re.sub(r'@\w+', '', texto)
    texto = re.sub(r'#\w+', '', texto)

    # Eliminamos todo lo que no sean letras, números o espacios
    texto = re.sub(r'[^\w\s]', '', texto)

    # Eliminamos espacios múltiples
    texto = re.sub(r'\s+', ' ', texto).strip()

    return texto

# Aplicamos función
corpus_limpio = [normalizar_texto(doc) for doc in corpus_es]

# Creamos los vectorizadores
count_vectorizer_limpio = CountVectorizer()
tfidf_vectorizer_limpio = TfidfVectorizer()

# Transformamos el corpus limpio
matriz_count_limpia = count_vectorizer_limpio.fit_transform(corpus_limpio)
matriz_tfidf_limpia = tfidf_vectorizer_limpio.fit_transform(corpus_limpio)

# Dimensiones
print("=== Dimensiones de las nuevas matrices ===")
print(f"Forma de la matriz CountVectorizer: {matriz_count_limpia.shape}")
print(f"Forma de la matriz TfidfVectorizer: {matriz_tfidf_limpia.shape}\n")

# Primeros 10 tokens y sus índices
print("=== Primeros 10 tokens y sus índices en el texto preprocesado ===")
vocabulario_limpio = count_vectorizer_limpio.vocabulary_
primeros_10_tokens_limpios = sorted(vocabulario_limpio.items(), key=lambda x: x[1])[:10]

for token, indice in primeros_10_tokens_limpios:
    print(f"Índice {indice}: {token}")
print("\n")

df_count_limpia = pd.DataFrame(
    matriz_count_limpia.toarray(),
    columns=count_vectorizer_limpio.get_feature_names_out()
)

df_tfidf_limpia = pd.DataFrame(
    matriz_tfidf_limpia.toarray(),
    columns=tfidf_vectorizer_limpio.get_feature_names_out()
)

print("=== Representación CountVectorizer Texto Limpio (Muestra de 5 filas x 10 columnas) ===")
print(df_count_limpia.iloc[:5, :10], "\n")

print("=== Representación TfidfVectorizer Texto Limpio (Muestra de 5 filas x 10 columnas) ===")
print(df_tfidf_limpia.iloc[:5, :10], "\n")
```

Diferencias observadas respecto al Ejercicio 1:
* Reducción de dimensionalidad: Las matrices tienen menos columbas debido a la eliminación de stopwords que no aportaban valor semántico.
* Los tokens ahora no contienen acentos, caracteres especiales ni URLs, lo que permite que el modelo trabaje con términos más estandarizados y evite tratar la misma palabra como dos distintas.
