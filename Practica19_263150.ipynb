# Ejercicio 01: Duplicidad parcial una columna: Contar y filtrar los registros duplicados en *Duration*

1. Carga el conjunto de datos dirtydata.csv y verifica que su carga sea correcta.
2. Cuenta la cantidad de registros duplicados utilizando solo la información de columna Duration e imprime
3. Utiliza value_counts() e imprime la cantidad de registros que hay por cada valor único.
4. Filtra los datos y obten los registros sin duplicados

```python
# 1. Carga el conjunto de datos dirtydata.csv y verifica que su carga sea correcta.
import pandas as pd
df = pd.read_csv('/content/drive/MyDrive/ClassFiles/dirtydata.csv')
print(f'Encabezado de la base de datos <dirtydata.csv>: \n{df.head()}')
print("*"*80)

# 2. Cuenta la cantidad de registros duplicados utilizando solo la información de columna Duration e imprime
registros_duplicados = df[df.duplicated(subset=['Duration'])]
print(f'Registros duplicados: \n{registros_duplicados}')
print("*"*80)

# 3. Utiliza value_counts() e imprime la cantidad de registros que hay por cada valor único.
print(f"Cantidad de registros por valor único: \n{registros_duplicados['Duration'].value_counts()}")
print("*"*80)

# 4. Filtra los datos y obten los registros sin duplicados
registros_sin_duplicados = df.drop_duplicates(subset=['Duration'])
print(f'Registros filtrados sin duplicados: \n {registros_sin_duplicados}')
```

# Ejercicio 02: Duplicidad parcial dos columnas: Contar y filtrar los registros duplicados en *Pulse* y *MaxPulse*

1. Carga el conjunto de datos dirtydata.csv y verifica que su carga sea correcta.
2. Cuenta la cantidad de registros duplicados utilizando la información de las columnas Pulse y Maxpulse
3. Utiliza value_counts() e imprime la cantidad de registros que hay por cada valor único.
4. Filtra los datos y obten los registros cuando keep = 'last'


```python
# 1. Carga el conjunto de datos dirtydata.csv y verifica que su carga sea correcta.
import pandas as pd
df2 = pd.read_csv('/content/drive/MyDrive/ClassFiles/dirtydata.csv')
print(f'Encabezado de la base de datos <dirtydata.csv>: \n{df2.head()}')
print("*"*80)

# 2. Cuenta la cantidad de registros duplicados utilizando la información de las columnas Pulse y Maxpulse
registros_duplicados_2 = df2[df2.duplicated(subset=['Pulse', 'Maxpulse'])]
print(f'Registros duplicados: \n{registros_duplicados_2}')
print("*"*80)

# 3. Utiliza value_counts() e imprime la cantidad de registros que hay por cada valor único.
valor_unico_pulse = registros_duplicados_2['Pulse'].value_counts()
valor_unico_maxpulse = registros_duplicados_2['Maxpulse'].value_counts()
print(f'Cantidad de registros por valor único en Pulse: \n{valor_unico_pulse}')
print(f'Cantidad de registros por valor único en Maxpulse: \n{valor_unico_maxpulse}')
print("*"*80)

conteo_combinaciones = registros_duplicados_2.groupby(['Pulse', 'Maxpulse']).size()
print(f'Cantidad de registros por cada combinación única duplicada: \n{conteo_combinaciones}')
print("*"*80)

# 4. Filtra los datos y obten los registros cuando keep = 'last'
registros_sin_duplicados_2 = df2.drop_duplicates(subset=['Pulse', 'Maxpulse'], keep='last')
print(f'Registros filtrados sin duplicados: \n {registros_sin_duplicados_2}')
```

# Ejercicio 03: Duplicidad total o exacta: Contar y filtrar los registros duplicados

1. Carga el conjunto de datos dirtydata.csv y verifica que su carga sea correcta.
2. Cuenta la cantidad de registros duplicados exactos (todas las columnas)
3. Utiliza sum() e imprime la cantidad de registros con duplicidad exacta
4. Filtra los datos y obten los registros cuando keep = 'last'


```python
# 1. Carga el conjunto de datos dirtydata.csv y verifica que su carga sea correcta.
import pandas as pd
df3 = pd.read_csv('/content/drive/MyDrive/ClassFiles/dirtydata.csv')
print(f'Encabezado de la base de datos <dirtydata.csv>: \n{df3.head()}')
print("*"*80)

# 2. Cuenta la cantidad de registros duplicados exactos (todas las columnas)
registros_duplicados_total = df3[df3.duplicated()]
print(f'Registros duplicados: \n{registros_duplicados_total}')
print("*"*80)

# 3. Utiliza sum() e imprime la cantidad de registros con duplicidad exacta
print(f'Cantidad de registros con duplicidad exacta: \n{df3.duplicated().sum()}')
print("*"*80)

# 4. Filtra los datos y obten los registros cuando keep = 'last'
registros_sin_duplicados_3 = df3.drop_duplicates(keep='last')
print(f'Registros filtrados sin duplicados: \n {registros_sin_duplicados_3}')
```

# Ejercicio 04: Cálculo de la media con duplicados y sin duplicados

1. Carga el conjunto de datos dirtydata.csv y verifica que su carga sea correcta.
2. Calcula e imprime la media de las columnas Pulse y MaxPulse
3. Identifica los valores duplicados en las columnas Pulse y MaxPulse (Duplicidad Parcial) y elimina los registro repetidos. Quédate con las últimas ocurrencias
4. Utiliza el conjunto anterior sin datos duplicados (recuerda que solo quedaron las últimas ocurrencias de los datos) y calcula la media de las columnas Pulse y MaxPulse


```python
# 1. Carga el conjunto de datos dirtydata.csv y verifica que su carga sea correcta.
import pandas as pd
df4 = pd.read_csv('/content/drive/MyDrive/ClassFiles/dirtydata.csv')
print(f'Encabezado de la base de datos <dirtydata.csv>: \n{df4.head()}')
print("*"*80)

# 2. Calcula e imprime la media de las columnas Pulse y MaxPulse
media_pulse = df4['Pulse'].mean()
media_maxpulse = df4['Maxpulse'].mean()
print(f'Media de Pulse (original): {media_pulse}')
print(f'Media de Maxpulse (original): {media_maxpulse}')
print("*"*80)

# 3. Identifica los valores duplicados en las columnas Pulse y MaxPulse (Duplicidad Parcial) y elimina los registro repetidos. Quédate con las últimas ocurrencias
valores_duplicados_4 = df4[df4.duplicated(subset=['Pulse', 'Maxpulse'], keep=False)]
print(f'Valores duplicados (todas las ocurrencias):\n{valores_duplicados_4}')
print("*"*80)

# 4. Utiliza el conjunto anterior sin datos duplicados (recuerda que solo quedaron las últimas ocurrencias de los datos) y calcula la media de las columnas Pulse y MaxPulse
df4_sin_duplicados = df4.drop_duplicates(subset=['Pulse', 'Maxpulse'], keep='last')
media_pulse_sin_duplicados2 = df4_sin_duplicados['Pulse'].mean()
media_maxpulse_sin_duplicados2 = df4_sin_duplicados['Maxpulse'].mean()
print(f'Media de Pulse (sin duplicados, keep=last): {media_pulse_sin_duplicados2}')
print(f'Media de Maxpulse (sin duplicados, keep=last): {media_maxpulse_sin_duplicados2}')
```
