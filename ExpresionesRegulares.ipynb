## Patrones comodín: caracteres especiales

Un punto (.) coincide con cualquier carácter, excepto el carácter de nueva línea.

```python
re.search(r'Co.k.e', 'Cookie').group()
```

Un signo de intercalación (^) coincide con el inicio de la cadena.

```python
re.search(r'^Eat', "Eat cake!").group()
```

El siguiente ejemplo no encuentra "eat" al inicio de la cadena, por eso devuelve None

```python
# However, the code below will not give the same result. Try it for yourself:
re.search(r'^eat', "Let's eat cake!").group()
```

$ - Coincide con el final de la cadena

```python
re.search(r'cake$', "Cake! Let's eat cake").group()
```

El siguiente ejemplo no encuentra "cake" al final de la cadena, por eso devuelve None

```python
## The next search will return the NONE value, try it:
re.search(r'cake$', "Let's get some cake on our way home!").group()
```

[abc] - Coincide con a o b o c. <br>
[a-zA-Z0-9] - Coincide con cualquier letra de (a a z) o (A a Z) o (0 a 9).

```python
re.search(r'[0-6]', 'Number: 5').group()
```

```python
## Matches any character except 5
re.search(r'Number: [^5]', 'Number: 0').group()
```

```python
## This will not match and hence a NONE value will be returned
re.search(r'Number: [^5]', 'Number: 5').group()
```

\ - Barra invertida.

1. Si el carácter que sigue a la barra invertida es un carácter de escape reconocido, entonces se toma el significado especial del término.

```python
## (Scenario 1) This treats '\s' as an escape character, '\s' defines a space
re.search(r'Not a\sregular character', 'Not a regular character').group()
```

\ - Barra invertida.<br>
2. Si el carácter que sigue a \ no es un carácter de escape reconocido, entonces \ se trata como cualquier otro carácter y se pasa.


```python
## (Scenario 2) '\' is treated as an ordinary character, because '\r' is not a recognized escape character
re.search(r'Just a \regular character', 'Just a \regular character').group()
```

\ - Barra invertida.<br>
3. \ puede utilizarse delante de todos los metacaracteres para eliminar su significado especial.

```python
## (Scenario 3) '\s' is escaped using an extra `\` so its interpreted as a literal string '\s'
re.search(r'Just a \\sregular character', 'Just a \sregular character').group()
```

Para evitar los *warnings* utilicé r en la segunda cadena:

```python
## (Scenario 3) '\s' is escaped using an extra `\` so its interpreted as a literal string '\s'
re.search(r'Just a \\sregular character', r'Just a \sregular character').group()
```

\w - La "w" minúscula. Coincide con cualquier letra, cifra o guión bajo.<br>
\W - W mayúscula. Coincide con cualquier carácter que no forme parte de \w (w minúscula).

```python
print("Lowercase w:", re.search(r'Co\wk\we', 'Cookie').group())

## Matches any character except single letter, digit or underscore
print("Uppercase W:", re.search(r'C\Wke', 'C@ke').group())

## Uppercase W won't match single letter, digit
print("Uppercase W won't match, and return:", re.search(r'Co\Wk\We', 'Cookie'))
```

\s - La "s" minúscula. Coincide con un único carácter de espacio en blanco como: espacio, nueva línea, tabulador, retorno.<br>
\S - S mayúscula. Coincide con cualquier carácter que no forme parte de \s (s minúscula).

```python
print("Lowercase s:", re.search(r'Eat\scake', 'Eat cake').group())
print("Uppercase S:", re.search(r'cook\Se', "Let's eat cookie").group())
```

\d - D minúscula. Coincide con el dígito decimal 0-9.<br>
\D - D mayúscula. Coincide con cualquier carácter que no sea un dígito decimal.

```python
# Example for \d
print("How many cookies do you want? ", re.search(r'\d+', '100 cookies').group())
```

* \t - T minúscula. Ficha Coincidencias.
* \n - N minúscula. Coincide con la nueva línea.
* \r - R minúscula. Devuelve las coincidencias.
* \A - Mayúscula a. Coincide sólo con el principio de la cadena. También funciona en varias líneas.
* \Z - Z mayúscula. Coincide sólo con el final de la cadena.
* \b - B minúscula. Sólo coincide con el principio o el final de la palabra.
* CONSEJO: ^ y \A son  lo mismo, y también lo son $ y \Z. Excepto cuando se trata del modo MULTILINEAL.

```python
# Example for \t
print("\\t (TAB) example: ", re.search(r'Eat\tcake', 'Eat\tcake').group())

# Example for \b
print("\\b match gives: ",re.search(r'\b[A-E]ookie', 'Cookie').group())
```

## Repeticiones

El signo más (+) comprueba si el caracter precedente aparece una o más veces a partir de esa posición.

```python
re.search(r'Co+kie', 'Cooookie').group()
```

El asterisco (*) comprueba si el caracter precedente aparece cero o más veces a partir de esa posición.

```python
# Checks for any occurrence of a or o or both in the given sequence
re.search(r'Ca*o*kie', 'Cookie').group()
```

El signo de cierre de interrogación (?) comprueba si el caracter precedente aparece exactamente cero o una vez a partir de esa posición.

```python
# Checks for exactly zero or one occurrence of a or o or both in the given sequence
re.search(r'Colou?r', 'Color').group()
```

{x} - Repite exactamente x número de veces.<br>
{x,} - Repite al menos x veces o más.<br>
{x, y} - Repite al menos x veces pero no más de y veces.

```python
re.search(r'\d{9,10}', '0987654321').group()
```

```python
re.search(r'\d{4,11}', '0987654321215648').group()
```

```python
re.search(r'\d{4,11}', '098765').group()
```

```python
re.search(r'\d{4,11}', '09').group()
```

## Agrupación en expresiones regulares

```python
statement = 'Please contact us at: support@datacamp.com'
match = re.search(r'([\w\.-]+)@([\w\.-]+)', statement)
if statement:
  print("Email address:", match.group()) # The whole matched text
  print("Username:", match.group(1)) # The username (group 1)
  print("Host:", match.group(2)) # The host (group 2)
```

```python
statement = 'Please contact us at: support@datacamp.com'
match = re.search(r'(?P<email>(?P<username>[\w\.-]+)@(?P<host>[\w\.-]+))', statement)
if statement:
  print("Email address:", match.group('email'))
  print("Username:", match.group('username'))
  print("Host:", match.group('host'))
```

## Codicioso vs. emparejamiento no arbitrario

```python
heading  = r'<h1>TITLE</h1>'
re.match(r'<.*>', heading).group()
```

Si añades ? después del calificador, la coincidencia se realizará de forma mínima, es decir, con el menor número posible de caracteres. Cuando ejecutes <.*>, sólo obtendrás una coincidencia con

```python
heading  = r'<h1>TITLE</h1>'
re.match(r'<.*?>', heading).group()
```

start() - Devuelve el índice inicial de la coincidencia.<br>
end() - Devuelve el índice donde termina la coincidencia.<br>
span() - Devuelve una tupla que contiene las posiciones (inicial, final) de la coincidencia.

```python
pattern = re.compile('COOKIE', re.IGNORECASE)
match = pattern.search("I am not a cookie monster")

print("Start index:", match.start())
print("End index:", match.end())
print("Tuple:", match.span())
```

## Banderas de compilación

IGNORECASE (I) - Permite coincidencias sin distinción entre mayúsculas y minúsculas.<br>
DOTALL (S) - Permite que . coincida con cualquier carácter, incluida la nueva línea.<br>
MULTILÍNEA (M) - Permite que el anclaje de inicio de cadena (^) y de fin de cadena ($) coincida también con nuevas líneas.<br>
VERBOSE (X) - Te permite escribir espacios en blanco y comentarios dentro de una expresión regular para hacerla más legible.

```python
statement = "Please contact us at: support@DataCamp.com, xyz@DATACAMP.com"

# Using the VERBOSE flag helps understand complex regular expressions
pattern = re.compile(r"""
[\w\.-]+ #First part
@ #Matches @ sign within email addresses
datacamp.com #Domain
""", re.X | re.I)

addresses = re.findall(pattern, statement)
for address in addresses:
    print("Address: ", address)
```

## Estudio de caso: Trabajar con expresiones regulares

```python
import re
import requests
the_idiot_url = 'https://www.gutenberg.org/files/2638/2638-0.txt'

def get_book(url):
    # Sends a http request to get the text from project Gutenberg
    raw = requests.get(url).text
    # Discards the metadata from the beginning of the book
    start = re.search(r"\*\*\* START OF THE PROJECT GUTENBERG EBOOK .* \*\*\*",raw ).end()
    # Discards the text starting Part 2 of the book
    stop = re.search(r"II", raw).start()
    # Keeps the relevant text
    text = raw[start:stop]
    return text

def preprocess(sentence):
    return re.sub('[^A-Za-z0-9.]+' , ' ', sentence).lower()

book = get_book(the_idiot_url)
processed_book = preprocess(book)
print(processed_book)
```

```python
# Encuentra el número del pronombre "the" en el corpus. Pista: Utiliza la función len().
def get_book2(url):
    raw = requests.get(url).text
    start = re.search(r"I\.", raw).end()
    text = raw[start:]
    return text

def preprocess2(sentence):
    return re.sub('[^A-Za-z0-9.]+', ' ', sentence).lower()

book2 = get_book2(the_idiot_url)
processed_book2 = preprocess2(book2)

count_the = len(re.findall(r'the', processed_book2))
print(f"La palabra 'the' aparece {count_the} veces")
```

```python
#  convertir cada instancia aislada de "i" en "I" en el corpus. Asegúrate de no cambiar la "i" que aparece dentro de una palabra:
def get_book3(url):
    raw = requests.get(url).text
    start = re.search(r"I\.", raw).end()
    stop = start + re.search(r"II\.", raw[start:]).start()
    text = raw[start:stop]
    return text

def preprocess3(sentence):
    sentence = re.sub(r'\bi\b', 'I', sentence, flags=re.IGNORECASE)
    sentence = re.sub(r'[^A-Za-z0-9.,?!;:\’\'"()\-]+', ' ', sentence)
    return sentence

book3 = get_book3(the_idiot_url)
processed_book3 = preprocess3(book3)

print(processed_book3[:5000])
```

```python
# Encuentra el número de veces que se citó a alguien ("") en el corpus

def get_book4(url):
    raw = requests.get(url).text
    start = re.search(r"I\.", raw).end()
    text = raw[start:]
    return text

book4 = get_book4(the_idiot_url)

quote_count = len(re.findall(r'”', book4))
print(f"Número de veces que se citó a alguien (”): {quote_count}")
```
