**Antes de ejecutar el siguiente código... Investiga:**

¿Por qué agregar la condición *if comunes.shape[1] == 0* ? Para quitar las películas que no calificaron, porque tienen valor 0

¿Qué valores tienen vector_1 y vector_2?
Son listas de Numpy, con las calificaciones de las películas de cada usuario.

¿Qué hace la instrucción *dot*?
Multiplica cada valor de las listas y luego los suma, es el producto punto.

¿Qué hace la instrucción *linalg.norm*?
Nos da como resultado la magnitud de un vector

```python
import numpy as np

def similitud_coseno_usuarios(matriz, usuario_1, usuario_2):

    comunes = matriz.loc[[usuario_1, usuario_2]].dropna(axis=1)

    if comunes.shape[1] == 0:
        return None

    vector_1 = comunes.loc[usuario_1].values
    vector_2 = comunes.loc[usuario_2].values

    producto_punto = np.dot(vector_1, vector_2)

    norma_1 = np.linalg.norm(vector_1)
    norma_2 = np.linalg.norm(vector_2)

    similitud = producto_punto / (norma_1 * norma_2)

    return similitud
```

## **Experimento A**
Cambiar el rating de Bruno para Misión orbital de 5 a 1.

```python
ratings_A = ratings.copy()

ratings_A.loc[
    (ratings_A["usuario"] == "Bruno") &
    (ratings_A["pelicula"] == "Misión orbital"),
    "rating"
] = 1

matriz_A = ratings_A.pivot_table(
    index="usuario",
    columns="pelicula",
    values="rating"
)

print(matriz_A)

sim_A = similitud_coseno_usuarios(matriz_A, "Ana", "Bruno")
print("Similitud Ana-Bruno (A):", sim_A)
```

## **Experimento B**

Eliminar una calificación de Diego.

```python
ratings_B = ratings.copy()

ratings_B = ratings_B.drop(
    ratings_B[
        (ratings_B["usuario"] == "Diego") &
        (ratings_B["pelicula"] == "Misión orbital")
    ].index
)

matriz_B = ratings_B.pivot_table(
    index="usuario",
    columns="pelicula",
    values="rating"
)

print(matriz_B)

sim_B = similitud_coseno_usuarios(matriz_B, "Ana", "Diego")
print("Similitud Ana-Diego (B):", sim_B)
```

## **Experimento C**

Agregar un usuario con una sola película en común con Ana.

```python
ratings_C = ratings.copy()

nuevo_usuario = pd.DataFrame({
    "usuario": ["Pancho"],
    "pelicula": ["Misión orbital"],
    "rating": [1]
})

ratings_C = pd.concat([ratings_C, nuevo_usuario], ignore_index=True)

matriz_C = ratings_C.pivot_table(
    index="usuario",
    columns="pelicula",
    values="rating"
)

print(matriz_C)

sim_C = similitud_coseno_usuarios(matriz_C, "Ana", "Pancho")
print("Similitud Ana-Pancho (C):", sim_C)
```
