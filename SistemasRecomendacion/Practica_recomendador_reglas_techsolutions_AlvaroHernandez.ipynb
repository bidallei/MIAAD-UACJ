## Etapa 3. Reglas obligatorias: determinar factibilidad

La función `evaluar_restricciones(solucion, usuario)` compara **un ítem** con **un perfil**. Cada `if` expresa una regla: si se viola una condición obligatoria, agregamos la causa a `razones` mediante `append()`.

Se evalúan **todas** las restricciones, para explicar más de una causa de exclusión. La lista vacía `[]` significa que **no se encontró ningún incumplimiento**.

### Antes de ejecutar: sintaxis y predicción

**Tu respuesta:** ...

13. `razones = []` inicia una lista vacía para almacenar los motivos por los que una solución no resulta viable. `razones.append('...')` agrega una nueva razón al final de esa lista cada vez que se detecta que no se cumple una regla.
14. El diccionario `solucion` nos da el valor del costo y el diccionario `usuario` nos da el límite de presupuesto definido en el perfil de la empresa.
15. `and` exige que ambas condiciones se cumplan al mismo tiempo para ejecutar el bloque de código y `not` es la negación de la condición. Juntos, verifican si la empresa requiere manejo de datos sensibles y no cuenta con una solución.
16. Solo puede con 50 usuarios y se requieren 180 usuarios; tiene disponibilidad de 99.0 y el mínimo es 99.9; y no maneja datos sensibles.

```python
def evaluar_restricciones(solucion, usuario):
    razones = []

    # R1. Presupuesto
    if solucion['costo'] > usuario['presupuesto']:
        razones.append('R1: Excede presupuesto')

    # R2. Capacidad
    if solucion['capacidad'] < usuario['usuarios']:
        razones.append('R2: Capacidad insuficiente')

    # R3. Disponibilidad
    if solucion['disponibilidad'] < usuario['disponibilidad_minima']:
        razones.append('R3: Disponibilidad insuficiente')

    # R4. Datos sensibles
    if usuario['maneja_datos_sensibles'] and not solucion['datos_sensibles']:
        razones.append('R4: No admite datos sensibles')

    # R5. Personal técnico
    if solucion['requiere_personal'] and not usuario['tiene_personal_tecnico']:
        razones.append('R5: Requiere personal técnico')

    return razones

print('Cloud Basic para Alfa:', evaluar_restricciones(SOLUCIONES[0], ALFA))
print('Cloud Business para Alfa:', evaluar_restricciones(SOLUCIONES[1], ALFA))
```

## Etapa 4. Reglas de preferencia: calcular puntuaciones

Las reglas R7–R12 **no excluyen** soluciones. Cuando una condición se satisface, se suman puntos a `puntos` y se guarda una explicación. El método `return puntos, explicaciones` devuelve **dos resultados**.

Los pesos son **ilustrativos**, no probabilidades y no fueron aprendidos de datos. Solo se comparan puntuaciones de soluciones que primero cumplieron las restricciones.

### Antes de ejecutar: sintaxis y predicción

**Tu respuesta:** ...


21. Se suman 20 puntos al valor actual de la variable. La expresión puntos = 20 solamente sobrescribe el valor existente y descarta cualquier acumulado previo.
22. Que las dos condiciones que se evalúan sean verdaderas para poder ejecutar el bloque de código.
23. La primera operación nos da el 80% del presupuesto de la empresa para poder bonificar las soluciones que viables económicamente. La segunda le da un 50% extra a la cantidad de usuarios actuales para premiar a las opciones que ofrezcan un buen margen.
24. De acuerdo a los datos, Cloud Business sumará puntos en dos reglas: en R8 por tener escalabilidad alta para un crecimiento alto (+25) y en R10 por ofrecer administración baja a una empresa sin personal técnico (+15).

```python
def calcular_puntuacion(solucion, usuario):
    puntos = 0
    explicaciones = []

    # R7. Tipo preferido
    if solucion['tipo'] == usuario['preferencia_tipo']:
        puntos += 20
        explicaciones.append('R7: Tipo preferido (+20)')

    # R8. Crecimiento y escalabilidad
    if usuario['crecimiento'] == 'alto' and solucion['escalabilidad'] == 'alta':
        puntos += 25
        explicaciones.append('R8: Alta escalabilidad (+25)')

    # R9. Confiabilidad
    if usuario['prioridad'] == 'confiabilidad' and solucion['disponibilidad'] >= 99.95:
        puntos += 20
        explicaciones.append('R9: Alta confiabilidad (+20)')

    # R10. Administración sencilla
    if solucion['administracion'] == 'baja' and not usuario['tiene_personal_tecnico']:
        puntos += 15
        explicaciones.append('R10: Administración sencilla (+15)')

    # R11. Precio favorable
    if solucion['costo'] <= usuario['presupuesto'] * 0.8:
        puntos += 10
        explicaciones.append('R11: Costo favorable (+10)')

    # R12. Capacidad adicional
    if solucion['capacidad'] >= usuario['usuarios'] * 1.5:
        puntos += 10
        explicaciones.append('R12: Capacidad adicional (+10)')

    return puntos, explicaciones

puntos, explicaciones = calcular_puntuacion(SOLUCIONES[1], ALFA)
print('Cloud Business para Alfa:', puntos, 'puntos')
for explicacion in explicaciones:
    print('-', explicacion)
```

## Etapa 5. Integrar y ordenar las recomendaciones

La función `recomendar()` recorre las soluciones con `for`. Para cada una, aplica **primero** las restricciones. Si hay razones, la guarda en `descartadas`; si no, calcula puntos y la guarda en `recomendaciones`.

Para mantener sencilla la sintaxis, utilizaremos una función auxiliar `obtener_puntos()` para ordenar sin expresiones `lambda`.

`sort(key=obtener_puntos, reverse=True)` ordena de **mayor a menor puntuación**. El corte `[:n]` conserva como máximo los primeros `n` elementos.

### Antes de ejecutar: sintaxis y predicción

32. ¿Cómo interpretas `recomendaciones[:n]` si `n=3`?
33. ¿Por qué la puntuación solo se calcula dentro del `else`?

**Tu respuesta:** ...

29. Itera cada solución para evaluarla
30. En Python, una lista con datos se evalúa como verdadera. Si se encontraron motivos de exclusión, la lista tiene texto y se ejecuta el primer bloque. Si la lista está vacía, se evalúa como falsa y el programa pasa al bloque alternativo para procesar la recomendación.
31. Toma el resultado del análisis de ese ciclo y lo guarda al final de la lista, guardando la información en un diccionario que contiene el nombre del ítem y sus resultados.
32. Son las primeras tres posiciones de las sugerencias.
33. Porque solo se evalúan las opciones factibles.

```python
def obtener_puntos(recomendacion):
    return recomendacion['puntos']


def recomendar(usuario, soluciones, n=3):
    recomendaciones = []
    descartadas = []

    for solucion in soluciones:
        razones = evaluar_restricciones(solucion, usuario)

        if razones:
            descartadas.append({
                'nombre': solucion['nombre'],
                'razones': razones
            })
        else:
            puntos, explicaciones = calcular_puntuacion(solucion, usuario)
            recomendaciones.append({
                'nombre': solucion['nombre'],
                'puntos': puntos,
                'explicaciones': explicaciones
            })

    recomendaciones.sort(key=obtener_puntos, reverse=True)
    return recomendaciones[:n], descartadas

recomendaciones_alfa, descartadas_alfa = recomendar(ALFA, SOLUCIONES)
print('Recomendaciones para Alfa:', len(recomendaciones_alfa))
print('Descartadas para Alfa:', len(descartadas_alfa))
```

```python
# Modificaciones para los retos
def evaluar_restricciones_reto(solucion, usuario):
    razones = []
    if solucion['costo'] > usuario['presupuesto']: razones.append('R1: Excede presupuesto')
    if solucion['capacidad'] < usuario['usuarios']: razones.append('R2: Capacidad insuficiente')
    if solucion['disponibilidad'] < usuario['disponibilidad_minima']: razones.append('R3: Disponibilidad insuficiente')
    if usuario['maneja_datos_sensibles'] and not solucion['datos_sensibles']: razones.append('R4: No admite datos sensibles')
    if solucion['requiere_personal'] and not usuario['tiene_personal_tecnico']: razones.append('R5: Requiere personal técnico')

    # Reto A: R6 Escalabilidad obligatoria
    if usuario.get('escalabilidad_obligatoria') and solucion['escalabilidad'] != 'alta':
        razones.append('R6: No tiene alta escalabilidad')

    return razones

def calcular_puntuacion_reto(solucion, usuario):
    puntos = 0
    explicaciones = []

    # Reto B: R7 modificada a 50
    if solucion['tipo'] == usuario['preferencia_tipo']:
        puntos += 50
        explicaciones.append('R7: Tipo preferido (+50)')

    if usuario['crecimiento'] == 'alto' and solucion['escalabilidad'] == 'alta':
        puntos += 25
        explicaciones.append('R8: Alta escalabilidad (+25)')

    if usuario['prioridad'] == 'confiabilidad' and solucion['disponibilidad'] >= 99.95:
        puntos += 20
        explicaciones.append('R9: Alta confiabilidad (+20)')

    if solucion['administracion'] == 'baja' and not usuario['tiene_personal_tecnico']:
        puntos += 15
        explicaciones.append('R10: Administración sencilla (+15)')

    if solucion['costo'] <= usuario['presupuesto'] * 0.8:
        puntos += 10
        explicaciones.append('R11: Costo favorable (+10)')

    if solucion['capacidad'] >= usuario['usuarios'] * 1.5:
        puntos += 10
        explicaciones.append('R12: Capacidad adicional (+10)')

    # Reto C: R13
    if solucion['tipo'] == 'nube' and not usuario['tiene_personal_tecnico']:
        puntos += 15
        explicaciones.append('R13: Nube sin personal (+15)')

    return puntos, explicaciones

def recomendar_reto(usuario, soluciones, n=3):
    recomendaciones = []
    descartadas = []
    for solucion in soluciones:
        razones = evaluar_restricciones_reto(solucion, usuario)
        if razones:
            descartadas.append({'nombre': solucion['nombre'], 'razones': razones})
        else:
            puntos, explicaciones = calcular_puntuacion_reto(solucion, usuario)
            recomendaciones.append({'nombre': solucion['nombre'], 'puntos': puntos, 'explicaciones': explicaciones})

    recomendaciones.sort(key=lambda r: r['puntos'], reverse=True)
    return recomendaciones[:n], descartadas

# Ejecución de los escenarios
alfa_reto = ALFA.copy()
alfa_reto['escalabilidad_obligatoria'] = True

print("--- Resultados Reto A (Alfa con escalabilidad obligatoria) ---")
rec_alfa, desc_alfa = recomendar_reto(alfa_reto, SOLUCIONES)
print("Factibles:", [(r['nombre'], r['puntos']) for r in rec_alfa])

print("\n--- Resultados Reto B y C (Beta original) ---")
rec_beta, desc_beta = recomendar_reto(BETA, SOLUCIONES)
print("Factibles Beta:", [(r['nombre'], r['puntos']) for r in rec_beta])
```
