# Ejercicio 01: Tipo de Datos Correcto

1. Carga el conjunto de datos airdata.csv y verifica que cada columna tenga el tipo de datos correcto
2. Si una columna de tipo temporal no está en el tipo de correcto realiza la conversión.
3. Usa dtypes para mostrar que cada columna, principalmente la que contiene datos temporales, está en el tipo de datos correcto
4. Imprime las primeras tres líneas

```python
# Importamos la librería de pandas y asignamos un alias (pd)
import pandas as pd

# Carga el conjunto de datos Airdata.csv y verifica que cada columna tenga el tipo de datos correcto
df = pd.read_csv("/content/drive/MyDrive/ClassFiles/Airdata.csv")
print("Primeras filas del dataset:")
print(df.head())
print("*"*100)
print("Tipos de datos iniciales:\n", df.dtypes, "\n")
print("*"*100)

# Si una columna de tipo temporal no está en el tipo de correcto realiza la conversión.
df_parser= pd.read_csv("/content/drive/MyDrive/ClassFiles/Airdata.csv", parse_dates=['DateTime'])

# Usa dtypes para mostrar que cada columna, principalmente la que contiene datos temporales, está en el tipo de datos correcto
print("Tipos de datos ya convertidos:\n", df_parser.dtypes, "\n")
print("*"*100)

# Imprime las primeras tres líneas
print("Primeras 3 filas:\n", df_parser.head(3))
```

# Ejercicio 02: Extraer componentes de fecha y tiempo

1. De la columna 'DateTime' extrae el año, mes, día y hora. Crea una nueva columna para cada componente que extraes
2. Imprime las primeras tres líneas del dataframe que ahora deberá tener 4 nuevas columnas.


```python
# De la columna 'DateTime' extrae el año, mes, día y hora. Crea una nueva columna para cada componente que extraes
pd_datetime = pd.to_datetime(df_parser['DateTime'])
df_parser['Year'] = pd_datetime.dt.year
df_parser['Month'] = pd_datetime.dt.month
df_parser['Day'] = pd_datetime.dt.day
df_parser['Hour'] = pd_datetime.dt.hour


# Imprime las primeras tres líneas del dataframe que ahora deberá tener 4 nuevas columnas.
print(df_parser.head())
```

# Ejercicio 03: Aritmética - Suma de Intervalos (timedelta)

1. Suma 10 días, 7 horas y 15 minutos a cada elemento de la columna 'DateTime'. Guarda el resultado en una nueva columna y agrega al dataframe.
2. Imprime la columna 'DateTime' y la columna que resulta de sumar 10 días, 7 horas y 15 minutos.


```python
# 1. Suma 10 días, 7 horas y 15 minutos a cada elemento de la columna 'DateTime'. Guarda el resultado en una nueva columna y agrega al dataframe.
columna_suma = pd_datetime + pd.Timedelta(days=10, hours=7, minutes=15)
df_parser['DateTime_suma'] = columna_suma

# 2. Imprime la columna 'DateTime' y la columna que resulta de sumar 10 días, 7 horas y 15 minutos.
print(df_parser[['DateTime', 'DateTime_suma']])
```

# Ejercicio 04: Timestamp

1. Convierte la columna 'DateTime' la cual tiene un formato de fecha legible a a timestamp Unix. Guarda el resultado en una nueva columna
2. Imprime las primeras tres filas de  la columna 'DateTime' y la columna que representa dicha fecha en timestamp Unix


```python
# 1. Convierte la columna 'DateTime' la cual tiene un formato de fecha legible a a timestamp Unix. Guarda el resultado en una nueva columna
columna_timestamp = pd.to_datetime(df_parser['DateTime']).astype(int) / 10**9
df_parser['DateTime_timestamp'] = columna_timestamp

# 2. Imprime las primeras tres filas de  la columna 'DateTime' y la columna que representa dicha fecha en timestamp Unix
print(df_parser[['DateTime', 'DateTime_timestamp']].head())
```
