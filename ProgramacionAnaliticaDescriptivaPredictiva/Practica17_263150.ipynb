# Ejercicio 01: Errores Tipográficos y Valores Inconsistentes: maritalstatus

1. Imprime de la columna 'maritalstatus' los valores únicos.
2. Imprime la frecuencia de cada valor único
3. Cambia el valor 'Never-married' a 'Single'
4. Imprime la frecuencia de cada valor único.


```python
# Importamos la librería de pandas y asignamos un alias (pd)
import pandas as pd

# Carga el archivo
df = pd.read_csv("/content/drive/MyDrive/ClassFiles/nls97.csv")

# Imprime de la columna 'maritalstatus' los valores únicos.
print("Valores únicos de la columna <maritalstatus>:\n", df['maritalstatus'].unique())
print('-'*60)

# Imprime la frecuencia de cada valor único
print("Frecuencia de valores:\n", df['maritalstatus'].value_counts())
print('-'*60)

# Cambia el valor 'Never-married' a 'Single'
df['maritalstatus'] = df['maritalstatus'].replace('Never-married', 'Single')
print("Valores únicos después del cambio <Never-married> a <Single>:\n", df['maritalstatus'].unique())
print('-'*60)

# Imprime la frecuencia de cada valor único.
print("Frecuencia de valores después del cambio <Never-married> a <Single>:\n", df['maritalstatus'].value_counts())
```

# Ejercicio 02: Errores Tipográficos y Valores Inconsistentes: highestdegree.

1. Imprime de la columna 'highestdegree' los valores únicos.
2. Imprime la frecuencia de cada valor único
3. Observa que cada valor categórico empieza con 0. Modifica las categorías para que todas estén en minúsculas y ninguna empiece con un número. Por ejemplo, si dice "2. High School", debería ser 'high school.
4. Los cambios se realizarán en la misma columna 'highestdegrees'.
5. Imprime los nuevos valores únicos.


```python
# Importamos la librería de pandas y asignamos un alias (pd)
import pandas as pd

# Carga el archivo
df = pd.read_csv("/content/drive/MyDrive/ClassFiles/nls97.csv")

# Imprime de la columna 'highestdegree' los valores únicos.
print("Valores únicos de la columna <highestdegree>:\n", df['highestdegree'].unique())
print('-'*60)

# Imprime la frecuencia de cada valor único
print("Frecuencia de valores:\n", df['highestdegree'].value_counts())
print('-'*60)

# Observa que cada valor categórico empieza con 0. Modifica las categorías para que
# todas estén en minúsculas y ninguna empiece con un número. Por ejemplo, si dice
# "2. High School", debería ser 'high school.
# Los cambios se realizarán en la misma columna 'highestdegrees'.
df['highestdegree'] = df['highestdegree'].str.replace(r'^\d+\.\s*', '', regex=True).str.lower()

# Imprime los nuevos valores únicos.
print("Valores únicos después del cambio:\n", df['highestdegree'].unique())
```

# Ejercicio 03: Alta Cardinalidad - Frecuencias Bajas

1. De la columna 'maritalstatus' une las categorías Widowed y 'Single' en una categoría más general llamada 'Single/No Partner'. Crea una nueva columna, 'marital_grouped' donde aparezcan las nuevas categorías de maritalstatus.
2. Imprime los valores únicos de la nueva columna 'marital_grouped'.
3. Imprime la frecuencia de los valores únicos de la nueva columna 'marital_grouped'


```python
# De la columna 'maritalstatus' une las categorías Widowed y 'Single' en una categoría más general llamada 'Single/No Partner'.
# Crea una nueva columna, 'marital_grouped' donde aparezcan las nuevas categorías de maritalstatus.
df['maritalstatus'] = df['maritalstatus'].replace('Never-married', 'Single')

df['marital_grouped'] = df['maritalstatus'].replace({'Single': 'Single/No Partner','Widowed': 'Single/No Partner'})

# Imprime los valores únicos de la nueva columna 'marital_grouped'.
print("Valores únicos de la nueva columna <marital_grouped>:\n", df['marital_grouped'].unique())
print('-'*60)

# Imprime la frecuencia de los valores únicos de la nueva columna 'marital_grouped'
print("Frecuencia de los valores únicos de la nueva columna <marital_grouped>:\n", df['marital_grouped'].value_counts())
```

**Ejercicio 04: Tipos de Datos**

1. Imprime el tipo de datos de la columna gender
2. Imprime la cantidad de bytes de dicha columna
3. Si la columna es de tipo object, modifica a category.
4. En caso de haber realizado la modificación del tipo de datos, imprime la cantidad de bytes que ocupa dicha columna con el nuevo tipo de datos

```python
import pandas as pd

df = pd.read_csv("/content/drive/MyDrive/ClassFiles/nls97.csv")

# Imprime el tipo de datos de la columna gender
print("Tipo de datos de la columna <gender>: ", df['gender'].dtype)
print('-'*60)

# Imprime la cantidad de bytes de dicha columna
print("Cantidad de bytes de la columna <gender>: ", df['gender'].memory_usage(deep=True))
print('-'*60)

# Si la columna es de tipo object, modifica a category.
df['gender'] = df['gender'].astype('category')

# En caso de haber realizado la modificación del tipo de datos, imprime la cantidad de bytes que ocupa dicha columna con el nuevo tipo de datos
print("Cantidad de bytes de la columna <gender> con el nuevo tipo de datos: ", df['gender'].memory_usage(deep=True))
```
